```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# Configurar semilla para reproducibilidad
np.random.seed(42)
random.seed(42)

print("📊 Generando dataset de 12 meses...")

# 1. Cargar datos actuales para obtener los CustomerIDs
df_actual = pd.read_csv('raw_data.csv')
clientes_unicos = df_actual['CustomerID'].unique()
print(f"✓ Clientes base: {len(clientes_unicos)}")

# 2. Listas para almacenar los datos generados
fechas = []
customer_ids = []
amounts = []
productos = []
order_ids = []
locations = []

# Fechas del período (12 meses)
fecha_inicio = datetime(2022, 6, 1)
fecha_fin = datetime(2023, 6, 10)

order_id = 890000  # ID inicial

# 3. Generar comportamiento para cada cliente
for i, cliente in enumerate(clientes_unicos):
    # Asignar perfil de cliente aleatorio
    rand = random.random()
    
    if rand < 0.15:  # 15% VIPs
        num_compras = random.randint(15, 30)
        monto_base = random.uniform(700, 1500)
        ubicaciones_fav = random.sample(['New York', 'London', 'Tokyo', 'Paris'], 2)
    elif rand < 0.40:  # 25% Frecuentes
        num_compras = random.randint(8, 14)
        monto_base = random.uniform(400, 800)
        ubicaciones_fav = random.sample(['New York', 'London', 'Tokyo', 'Paris'], 2)
    elif rand < 0.70:  # 30% Ocasionales
        num_compras = random.randint(3, 7)
        monto_base = random.uniform(200, 500)
        ubicaciones_fav = ['New York', 'London', 'Tokyo', 'Paris']
    else:  # 30% Nuevos/Esporádicos
        num_compras = random.randint(1, 3)
        monto_base = random.uniform(100, 300)
        ubicaciones_fav = ['New York', 'London', 'Tokyo', 'Paris']
    
    # Generar fechas distribuidas en el año
    for _ in range(num_compras):
        # Fecha aleatoria dentro del período
        dias_transcurridos = (fecha_fin - fecha_inicio).days
        fecha_random = fecha_inicio + timedelta(days=random.randint(0, dias_transcurridos))
        
        # Monto con variación
        monto = monto_base * random.uniform(0.6, 1.4)
        
        fechas.append(fecha_random.strftime('%Y-%m-%d'))
        customer_ids.append(cliente)
        amounts.append(round(monto, 2))
        productos.append(random.choice(['Product A', 'Product B', 'Product C', 'Product D']))
        order_ids.append(order_id)
        locations.append(random.choice(ubicaciones_fav))
        
        order_id += 1

# 4. Crear DataFrame
df_extendido = pd.DataFrame({
    'CustomerID': customer_ids,
    'PurchaseDate': fechas,
    'TransactionAmount': amounts,
    'ProductInformation': productos,
    'OrderID': order_ids,
    'Location': locations
})

# 5. Ordenar por fecha
df_extendido = df_extendido.sort_values('PurchaseDate').reset_index(drop=True)

# 6. Guardar archivo
df_extendido.to_csv('raw_data_12meses.csv', index=False)

# 7. Mostrar estadísticas
print("\n" + "="*60)
print("✅ DATASET GENERADO EXITOSAMENTE")
print("="*60)
print(f"📁 Archivo: raw_data_12meses.csv")
print(f"📊 Total transacciones: {len(df_extendido):,}")
print(f"👥 Clientes únicos: {df_extendido['CustomerID'].nunique()}")
print(f"📅 Período: {df_extendido['PurchaseDate'].min()} a {df_extendido['PurchaseDate'].max()}")
print(f"💰 Monto promedio: ${df_extendido['TransactionAmount'].mean():.2f}")
print(f"🛒 Compras promedio por cliente: {df_extendido.groupby('CustomerID').size().mean():.1f}")
print("="*60)

# 8. Ver primeras filas
print("\n📋 Primeras 10 filas:")
print(df_extendido.head(10))
```
